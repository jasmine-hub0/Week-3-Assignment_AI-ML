```python
# Inventory Management System

# Task 1: Create inventory
inventory = {
    "Widget": 10,
    "Gadget": 5,
    "Sensor": 0,
    "Cable": 15
}

# Create order queue
orders = [
    ["Widget", 3],
    ["Sensor", 2],
    ["Gadget", 7],
    ["Cable", 5],
    ["Mouse", 2]
]

# Task 2: Process the orders
unfulfilled = []
fully_fulfilled = 0

for order in orders:
    item = order[0]
    quantity = order[1]

    # Check if item is in inventory
    if item not in inventory:
        print(item, "is not available in inventory.")
        unfulfilled.append([item, quantity])

    else:
        stock = inventory.get(item)

        # Full fulfillment
        if stock >= quantity:
            inventory[item] = stock - quantity
            fully_fulfilled += 1
            print(item, "order fully fulfilled.")

        # Partial fulfillment
        elif stock > 0:
            remaining = quantity - stock
            inventory[item] = 0
            unfulfilled.append([item, remaining])

            print(item, "partially fulfilled.")
            print("Remaining quantity:", remaining)

        # Out of stock
        else:
            print(item, "is out of stock.")
            unfulfilled.append([item, quantity])

# Task 3: Display final summary
print("\nFinal Inventory:")
print(inventory)

print("Fully fulfilled orders:", fully_fulfilled)

print("Unfulfilled orders:")
print(unfulfilled)
```
